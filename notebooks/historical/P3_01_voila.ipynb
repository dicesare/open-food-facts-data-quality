```python
data = pa.read_csv('open_food_fact_clean', sep='\t',dtype={'code':object})
data.shape
```

```python

from mpl_toolkits.mplot3d import Axes3D
output = widgets.Output()
```

```python
X_df = data[['fat_100g','carbohydrates_100g','fiber_100g','proteins_100g','salt_100g','nutrition_score_fr_100g']].dropna()
X_df_score = X_df.join(data.nutrition_grade_fr)
nutri = X_df_score.nutrition_grade_fr.astype('category')
Y = nutri.cat.codes
```

```python
X = scale(X_df.dropna())
pca_n = PCA()
pca_n.fit(X)
X_projected = pca_n.transform(X)
X_nutri = pa.DataFrame(X_projected,index=X_df.index).join(data.nutrition_grade_fr)
```

```python
# X_pivot = X_nutri.pivot_table(index='nutrition_grade_fr')
X_a = pca_n.transform(X_nutri[X_nutri.nutrition_grade_fr=='a'].drop(columns=['nutrition_grade_fr']))
X_b = pca_n.transform(X_nutri[X_nutri.nutrition_grade_fr=='b'].drop(columns=['nutrition_grade_fr']))
X_c = pca_n.transform(X_nutri[X_nutri.nutrition_grade_fr=='c'].drop(columns=['nutrition_grade_fr']))
X_d = pca_n.transform(X_nutri[X_nutri.nutrition_grade_fr=='d'].drop(columns=['nutrition_grade_fr']))
X_e = pca_n.transform(X_nutri[X_nutri.nutrition_grade_fr=='e'].drop(columns=['nutrition_grade_fr']))
```
