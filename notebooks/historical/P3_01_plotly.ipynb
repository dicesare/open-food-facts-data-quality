```python
df = pa.read_csv('open_food_fact_clean', sep=';',dtype={'code':object})
df.shape
```

```python

columns_select=['fat_100g','carbohydrates_100g','fiber_100g','proteins_100g','salt_100g','nutrition_score_fr_100g']
X_df = df[columns_select].copy().dropna()
X = X_df.copy()
```

```python
X = scale(X)
```

```python
X_df_pca = pa.DataFrame(X,columns=X_df.columns.values,index=X_df.index.values)
```

```python
mypca = PCA(n_components=3)
mypca.fit(X)
```

```python
X_scaled = mypca.fit_transform(X)
```

```python
X_projected = mypca.transform(X)
X_scaled_df = pa.DataFrame(X_scaled,columns=X_df.iloc[:,[0,1,2]].columns.values,index=X_df.index.values).join([df.additives_n,df.nutrition_grade_fr,df.nutrition_score_fr_100g])
X_projected_df = pa.DataFrame(X_projected,columns=X_df.iloc[:,[0,1,2]].columns.values,index=X_df.index.values).join([df.additives_n,df.nutrition_grade_fr,df.nutrition_score_fr_100g])
```
